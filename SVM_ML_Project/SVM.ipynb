```python
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

```python
df = datasets.load_iris(as_frame = True).frame
```

```python
X = df.drop("target",axis =1)
y = df["target"]
```

```python
X_test,X_train,y_test,y_train= train_test_split(
    X,y,
    test_size=0.3,
    random_state= 42,
    stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled =scaler.transform(X_test)
```

```python
svc = SVC()
svc.fit(X_train_scaled,y_train)
```

```python
y_pred = svc.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report
print("accuracy", accuracy_score(y_test,y_pred))
print("classification report\n",classification_report(y_test,y_pred))
```

```python
svc = SVC(kernel = "linear")
svc.fit(X_train_scaled,y_train)
y_pred = svc.predict(X_test_scaled)
print("accuracy", accuracy_score(y_test,y_pred))
```

```python
svc = SVC(kernel = "poly")
svc.fit(X_train_scaled,y_train)
y_pred = svc.predict(X_test_scaled)
print("accuracy", accuracy_score(y_test,y_pred))
```

```python
svc = SVC(kernel = "sigmoid")
svc.fit(X_train_scaled,y_train)
y_pred = svc.predict(X_test_scaled)
print("accuracy", accuracy_score(y_test,y_pred))
```

```python
C_vals = [0.5,1,2,3,4,5]
for c_vals in C_vals:
    svc = SVC(C=c_vals,kernel ="rbf")
    svc.fit(X_train_scaled,y_train)
    y_pred = svc.predict(X_test_scaled)
    print("c=",c_vals,"accuracy", accuracy_score(y_test,y_pred))
```

```python
from sklearn.svm  import SVR
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

```python
df =datasets.load_iris(as_frame= True).frame
```

```python
X = df.drop("target",axis=1)
y = df["target"]
```

```python
X_test,X_train,y_test,y_train= train_test_split(
    X,y,
    test_size=0.3,
    random_state= 42,
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
y_train_scaled = scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled =scaler.transform(y_test.values.reshape(-1,1)).ravel()
```

```python
model =SVR()
model.fit(X_train,y_train_scaled)
```

```python
y_pred_scaled =model.predict(X_test)
y_trainp_scaled =model.predict(X_train)
```

```python
from sklearn.metrics import r2_score
print(r2_score(y_test_scaled,y_pred_scaled))
print(r2_score(y_train_scaled,y_trainp_scaled))
```

```python
from sklearn.model_selection import GridSearchCV
para_grid = {
    "C" : [1,2,5,10,50,100],
    "kernel": ["rbf","linear"],
    "epsilon":[0.01,0.1,0.2,0.3,0.5]
}
```

```python
svr =SVR()
grid_search =GridSearchCV(svr,para_grid,scoring="r2",cv=5)
grid_search.fit(X_train,y_train_scaled)
```

```python
print("best param-",grid_search.best_params_)
```

```python
model =SVR(kernel = "linear",C=10,epsilon =0.1)
model.fit(X_train,y_train_scaled)
y_pred_scaled =model.predict(X_test)
y_trainp_scaled =model.predict(X_train)
```

```python
print(r2_score(y_test_scaled,y_pred_scaled))
print(r2_score(y_train_scaled,y_trainp_scaled))
```

```python
from sklearn.svm import LinearSVR
model = LinearSVR(C=10,epsilon=0.1,max_iter = 2000000)
model.fit(X_train,y_train_scaled)
y_pred_scaled =model.predict(X_test)
y_trainp_scaled =model.predict(X_train)
```

```python
print(r2_score(y_test_scaled,y_pred_scaled))
print(r2_score(y_train_scaled,y_trainp_scaled))
```
